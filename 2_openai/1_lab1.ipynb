```python
# The imports
import os
from dotenv import load_dotenv
from openai import AsyncOpenAI
from agents import Agent, Runner, trace, set_default_openai_api, set_default_openai_client

```

```python
# The usual starting point

load_dotenv(override=True)
```

```python

# Make an agent with name, instructions, model
google_api_key = os.getenv('GEMINI_API_KEY')
google_base_url = os.getenv('GEMINI_BASE_URL')

gemini_client = AsyncOpenAI(api_key= google_api_key, base_url=google_base_url)

set_default_openai_client(gemini_client)
set_default_openai_api("chat_completions")


agent = Agent(
    name="Joker",
    instructions="You are a joke teller.",
    model="gemini-flash-latest",
)
```

```python
agent
```

```python
# Run the joke with Runner.run(agent, prompt) then print final_output

result = await Runner.run(agent, "Tell a joke about Autonomous AI Agents")
print(result.final_output)
```
